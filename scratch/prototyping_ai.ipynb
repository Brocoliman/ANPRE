```python
import paddle
from paddleocr import PaddleOCR, TableRecognitionPipelineV2
paddle.utils.run_check()             # should print success
ocr = PaddleOCR()                    # no errors now
print("✔ PaddleOCR ready")
```

```python


pipeline = TableRecognitionPipelineV2(
    device="cpu",                         
    use_doc_orientation_classify=False,
    use_doc_unwarping=False
)

path = 'test_patient_info.png'
outputs = pipeline.predict(path)

# iterate over each table found in the page
for i, res in enumerate(outputs):
    print(f"\n=== Table {i} ===")
    # pretty-print the JSON result to console
    res.print()  

    # save visualizations & structured data
    res.save_to_img(f"./out/table_{i}.png")       # annotated image
    res.save_to_xlsx(f"./out/table_{i}.xlsx")     # Excel file
    res.save_to_html(f"./out/table_{i}.html")     # HTML table
    res.save_to_json(f"./out/table_{i}.json")     # JSON with cell coords & texts




```
