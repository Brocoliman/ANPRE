```python
import numpy as np
import fitz
import matplotlib.pyplot as plt
import cv2
```

```python
path = '../test_patient_info.pdf'

fitz_pdf = fitz.open(path)

page1, page2 = fitz_pdf[0], fitz_pdf[1]
page1 = page1.get_pixmap(dpi=150)
page2 = page2.get_pixmap(dpi=150)

img1 = np.frombuffer(page1.samples, dtype=np.uint8).reshape(page1.height, page1.width, page1.n)
img2 = np.frombuffer(page2.samples, dtype=np.uint8).reshape(page2.height, page2.width, page2.n)

plt.figure(figsize=(5, 5))
f, ax = plt.subplots(1,2) 
ax[0].imshow(img1)
ax[1].imshow(img2)

cv2.imwrite("../test_patient_info.png", img1)
```
